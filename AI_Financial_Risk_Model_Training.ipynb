```python
# ============================================================
# AI Financial Risk Analytics Platform
# Stage 4 : Data Cleaning
# ============================================================

import pandas as pd
import numpy as np
import time

print("=" * 70)
print("STAGE 4 : DATA CLEANING")
print("=" * 70)

# -----------------------------------------------------------
# Start Timer
# -----------------------------------------------------------
start_time = time.time()

# -----------------------------------------------------------
# Dataset Before Cleaning
# -----------------------------------------------------------
print("\nDataset Before Cleaning")
print("-" * 50)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

# -----------------------------------------------------------
# Calculate Missing Percentage
# -----------------------------------------------------------
missing_percentage = (df.isnull().sum() / len(df)) * 100

# -----------------------------------------------------------
# Columns having more than 65% Missing Values
# -----------------------------------------------------------
threshold = 65

drop_columns = missing_percentage[missing_percentage > threshold].index.tolist()

print("\nColumns to Drop (>65% Missing Values)")
print("-" * 50)

for col in drop_columns:
    print(col)

print(f"\nTotal Columns to Drop : {len(drop_columns)}")

# -----------------------------------------------------------
# Drop Columns
# -----------------------------------------------------------
df.drop(columns=drop_columns, inplace=True)

print("\nColumns Dropped Successfully")

# -----------------------------------------------------------
# Numerical Columns
# -----------------------------------------------------------
numerical_columns = df.select_dtypes(include=['int64', 'float64']).columns

# Fill Numerical Missing Values with Median
for col in numerical_columns:

    if df[col].isnull().sum() > 0:

        median_value = df[col].median()

        df[col].fillna(median_value, inplace=True)

# -----------------------------------------------------------
# Categorical Columns
# -----------------------------------------------------------
categorical_columns = df.select_dtypes(include=['object']).columns

# Fill Categorical Missing Values with Mode
for col in categorical_columns:

    if df[col].isnull().sum() > 0:

        mode_value = df[col].mode()[0]

        df[col].fillna(mode_value, inplace=True)

# -----------------------------------------------------------
# Missing Values After Cleaning
# -----------------------------------------------------------
remaining_missing = df.isnull().sum().sum()

print("\nRemaining Missing Values")
print("-" * 50)

print(remaining_missing)

# -----------------------------------------------------------
# Dataset After Cleaning
# -----------------------------------------------------------
print("\nDataset After Cleaning")
print("-" * 50)

print(f"Rows    : {df.shape[0]:,}")
print(f"Columns : {df.shape[1]}")

# -----------------------------------------------------------
# End Timer
# -----------------------------------------------------------
end_time = time.time()

print("\nCleaning Completed Successfully")

print(f"\nTime Taken : {end_time-start_time:.2f} seconds")

# -----------------------------------------------------------
# Save Clean Dataset
# -----------------------------------------------------------
OUTPUT_PATH = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/Datasets/application_train_clean.csv"

df.to_csv(OUTPUT_PATH, index=False)

print("\nClean Dataset Saved Successfully")

print(OUTPUT_PATH)
```

```python
# ============================================================
# Stage 7 : Feature Engineering
# ============================================================

import numpy as np
import pandas as pd
import time

print("=" * 70)
print("STAGE 7 : FEATURE ENGINEERING")
print("=" * 70)

start_time = time.time()

# -----------------------------------------------------------
# Function to Create Features
# -----------------------------------------------------------

def create_features(df):

    df = df.copy()

    # Credit to Income Ratio
    df["CREDIT_INCOME_RATIO"] = (
        df["AMT_CREDIT"] /
        df["AMT_INCOME_TOTAL"].replace(0, np.nan)
    )

    # Annuity to Income Ratio
    df["ANNUITY_INCOME_RATIO"] = (
        df["AMT_ANNUITY"] /
        df["AMT_INCOME_TOTAL"].replace(0, np.nan)
    )

    # Goods Price to Credit Ratio
    df["GOODS_CREDIT_RATIO"] = (
        df["AMT_GOODS_PRICE"] /
        df["AMT_CREDIT"].replace(0, np.nan)
    )

    # Employment in Years
    df["EMPLOYMENT_YEARS"] = abs(df["DAYS_EMPLOYED"]) / 365

    # Age in Years
    df["AGE_YEARS"] = abs(df["DAYS_BIRTH"]) / 365

    # Replace Inf values
    df.replace([np.inf, -np.inf], np.nan, inplace=True)

    # Fill any newly created NaNs using median
    new_features = [
        "CREDIT_INCOME_RATIO",
        "ANNUITY_INCOME_RATIO",
        "GOODS_CREDIT_RATIO",
        "EMPLOYMENT_YEARS",
        "AGE_YEARS"
    ]

    for col in new_features:
        df[col] = df[col].fillna(df[col].median())

    return df

# -----------------------------------------------------------
# Shapes Before Feature Engineering
# -----------------------------------------------------------

print("\nBefore Feature Engineering")
print("-" * 40)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)

# -----------------------------------------------------------
# Apply Feature Engineering
# -----------------------------------------------------------

X_train = create_features(X_train)
X_test = create_features(X_test)

# -----------------------------------------------------------
# Shapes After Feature Engineering
# -----------------------------------------------------------

print("\nAfter Feature Engineering")
print("-" * 40)

print("X_train :", X_train.shape)
print("X_test  :", X_test.shape)

# -----------------------------------------------------------
# Verify Missing Values
# -----------------------------------------------------------

print("\nMissing Values in Engineered Features")
print("-" * 40)

engineered_features = [
    "CREDIT_INCOME_RATIO",
    "ANNUITY_INCOME_RATIO",
    "GOODS_CREDIT_RATIO",
    "EMPLOYMENT_YEARS",
    "AGE_YEARS"
]

print(X_train[engineered_features].isnull().sum())

# -----------------------------------------------------------
# Save Updated Datasets
# -----------------------------------------------------------

train_engineered = X_train.copy()
train_engineered["TARGET"] = y_train.values

test_engineered = X_test.copy()
test_engineered["TARGET"] = y_test.values

train_path = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/Datasets/train_data_engineered.csv"
test_path = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/Datasets/test_data_engineered.csv"

train_engineered.to_csv(train_path, index=False)
test_engineered.to_csv(test_path, index=False)

end_time = time.time()

print("\nDatasets Saved Successfully")
print(train_path)
print(test_path)

print(f"\nTime Taken : {end_time - start_time:.2f} seconds")
```

```python
# ============================================================
# Stage 8 : Data Preprocessing Pipeline
# ============================================================

import time

from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import OneHotEncoder
import joblib

print("=" * 70)
print("STAGE 8 : DATA PREPROCESSING PIPELINE")
print("=" * 70)

start_time = time.time()

# -----------------------------------------------------------
# Identify Numerical and Categorical Columns
# -----------------------------------------------------------

numerical_features = X_train.select_dtypes(
    include=["int64", "float64"]
).columns.tolist()

categorical_features = X_train.select_dtypes(
    include=["object", "category"]
).columns.tolist()

print("\nNumber of Numerical Features :", len(numerical_features))
print("Number of Categorical Features :", len(categorical_features))

# -----------------------------------------------------------
# Numerical Pipeline
# -----------------------------------------------------------

numeric_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="median"))
    ]
)

# -----------------------------------------------------------
# Categorical Pipeline
# -----------------------------------------------------------

categorical_pipeline = Pipeline(
    steps=[
        ("imputer", SimpleImputer(strategy="most_frequent")),
        (
            "encoder",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=False
            )
        )
    ]
)

# -----------------------------------------------------------
# Combine Pipelines
# -----------------------------------------------------------

preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_pipeline, numerical_features),
        ("cat", categorical_pipeline, categorical_features)
    ],
    remainder="drop"
)

# -----------------------------------------------------------
# Fit on Training Data Only
# -----------------------------------------------------------

X_train_processed = preprocessor.fit_transform(X_train)

# -----------------------------------------------------------
# Transform Test Data
# -----------------------------------------------------------

X_test_processed = preprocessor.transform(X_test)

# -----------------------------------------------------------
# Verification
# -----------------------------------------------------------

print("\nProcessed Dataset Shapes")
print("-" * 40)

print("X_train_processed :", X_train_processed.shape)
print("X_test_processed  :", X_test_processed.shape)

import numpy as np

print("\nContains Missing Values?")
print("-" * 40)

print("Training :", np.isnan(X_train_processed).sum())
print("Testing  :", np.isnan(X_test_processed).sum())
# -----------------------------------------------------------
# Save Preprocessing Pipeline
# -----------------------------------------------------------

import os

# Create models folder if it doesn't exist
model_dir = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/models"
os.makedirs(model_dir, exist_ok=True)

pipeline_path = os.path.join(model_dir, "preprocessor.pkl")

joblib.dump(preprocessor, pipeline_path)

print("\nPreprocessing Pipeline Saved Successfully")
print(pipeline_path)

print("\nPreprocessing Pipeline Saved Successfully")

print(pipeline_path)

end_time = time.time()

print(f"\nTime Taken : {end_time-start_time:.2f} seconds")
```

```python
# ============================================================
# Stage 9 : Model Training
# ============================================================

import time
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score
)

print("="*70)
print("STAGE 9 : MODEL TRAINING")
print("="*70)

# -----------------------------------------------------------
# Models
# -----------------------------------------------------------

models = {
    "Logistic Regression": LogisticRegression(
        max_iter=1000,
        random_state=42
    ),

    "Decision Tree": DecisionTreeClassifier(
        random_state=42
    ),

    "Random Forest": RandomForestClassifier(
        n_estimators=100,
        random_state=42,
        n_jobs=-1
    ),

    "XGBoost": XGBClassifier(
        random_state=42,
        eval_metric="logloss"
    ),

    "LightGBM": LGBMClassifier(
        random_state=42
    ),

    "CatBoost": CatBoostClassifier(
        random_state=42,
        verbose=0
    )
}

# -----------------------------------------------------------
# Training Loop
# -----------------------------------------------------------

results = []

for name, model in models.items():

    print("\n" + "="*60)
    print(f"Training : {name}")
    print("="*60)

    start = time.time()

    model.fit(X_train_processed, y_train)

    end = time.time()

    train_time = end - start

    predictions = model.predict(X_test_processed)

    probabilities = model.predict_proba(X_test_processed)[:,1]

    accuracy = accuracy_score(y_test, predictions)

    precision = precision_score(
        y_test,
        predictions,
        zero_division=0
    )

    recall = recall_score(
        y_test,
        predictions,
        zero_division=0
    )

    f1 = f1_score(
        y_test,
        predictions,
        zero_division=0
    )

    roc_auc = roc_auc_score(
        y_test,
        probabilities
    )

    results.append([
        name,
        accuracy,
        precision,
        recall,
        f1,
        roc_auc,
        train_time
    ])

    print(f"Accuracy : {accuracy:.4f}")
    print(f"ROC AUC  : {roc_auc:.4f}")
    print(f"Time     : {train_time:.2f} sec")

# -----------------------------------------------------------
# Results Table
# -----------------------------------------------------------

results_df = pd.DataFrame(
    results,
    columns=[
        "Model",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score",
        "ROC AUC",
        "Training Time (sec)"
    ]
)

results_df = results_df.sort_values(
    by="ROC AUC",
    ascending=False
).reset_index(drop=True)

print("\n")
print("="*70)
print("MODEL COMPARISON")
print("="*70)

print(results_df)

# -----------------------------------------------------------
# Save Results
# -----------------------------------------------------------

results_path = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/reports/model_comparison.csv"

import os
os.makedirs(os.path.dirname(results_path), exist_ok=True)

results_df.to_csv(results_path, index=False)

print("\nResults saved successfully.")
print(results_path)
```

```python
# ============================================================
# Stage 10 : Hyperparameter Tuning
# ============================================================

import time
import pandas as pd

from sklearn.model_selection import RandomizedSearchCV

from xgboost import XGBClassifier
from lightgbm import LGBMClassifier
from catboost import CatBoostClassifier

from sklearn.metrics import roc_auc_score

print("="*70)
print("STAGE 10 : HYPERPARAMETER TUNING")
print("="*70)

# -----------------------------------------------------------
# Models and Parameter Grids
# -----------------------------------------------------------

models = {

    "LightGBM": (
        LGBMClassifier(random_state=42),

        {
            "n_estimators":[100,200,300],
            "learning_rate":[0.01,0.05,0.1],
            "max_depth":[3,5,7,-1],
            "num_leaves":[31,50,70]
        }

    ),

    "XGBoost": (

        XGBClassifier(
            random_state=42,
            eval_metric="logloss"
        ),

        {
            "n_estimators":[100,200,300],
            "learning_rate":[0.01,0.05,0.1],
            "max_depth":[3,5,7],
            "subsample":[0.8,1.0]
        }

    ),

    "CatBoost": (

        CatBoostClassifier(
            verbose=0,
            random_state=42
        ),

        {
            "iterations":[200,300],
            "learning_rate":[0.03,0.05,0.1],
            "depth":[4,6,8]
        }

    )

}

results = []

best_models = {}

# -----------------------------------------------------------
# Random Search
# -----------------------------------------------------------

for name, (model, params) in models.items():

    print("\n"+"="*60)
    print(f"Tuning : {name}")
    print("="*60)

    start = time.time()

    random_search = RandomizedSearchCV(

        estimator=model,

        param_distributions=params,

        n_iter=10,

        scoring="roc_auc",

        cv=3,

        verbose=1,

        random_state=42,

        n_jobs=-1

    )

    random_search.fit(X_train_processed, y_train)

    end = time.time()

    best_model = random_search.best_estimator_

    probabilities = best_model.predict_proba(X_test_processed)[:,1]

    roc_auc = roc_auc_score(y_test, probabilities)

    results.append([

        name,

        roc_auc,

        random_search.best_params_,

        end-start

    ])

    best_models[name] = best_model

    print(f"\nBest ROC-AUC : {roc_auc:.4f}")

# -----------------------------------------------------------
# Results
# -----------------------------------------------------------

results_df = pd.DataFrame(

    results,

    columns=[

        "Model",

        "ROC AUC",

        "Best Parameters",

        "Time (sec)"

    ]

)

results_df = results_df.sort_values(

    by="ROC AUC",

    ascending=False

).reset_index(drop=True)

print("\n")
print("="*70)
print("TUNING RESULTS")
print("="*70)

print(results_df)

# -----------------------------------------------------------
# Save Results
# -----------------------------------------------------------

import os

os.makedirs(
    "/content/drive/MyDrive/AI_Financial_Risk_Analytics/reports",
    exist_ok=True
)

results_df.to_csv(
    "/content/drive/MyDrive/AI_Financial_Risk_Analytics/reports/hyperparameter_results.csv",
    index=False
)

print("\nResults Saved Successfully.")
```

```python
# ============================================================
# Stage 11 : Final Model Evaluation
# ============================================================

import matplotlib.pyplot as plt
from sklearn.metrics import (
    ConfusionMatrixDisplay,
    classification_report,
    roc_curve,
    auc,
    precision_recall_curve,
    average_precision_score
)

print("=" * 70)
print("STAGE 11 : FINAL MODEL EVALUATION")
print("=" * 70)

# -----------------------------------------------------------
# Select Best Model
# -----------------------------------------------------------

best_model = best_models["CatBoost"]

# -----------------------------------------------------------
# Predictions
# -----------------------------------------------------------

y_pred = best_model.predict(X_test_processed)
y_prob = best_model.predict_proba(X_test_processed)[:, 1]

# -----------------------------------------------------------
# Classification Report
# -----------------------------------------------------------

print("\nClassification Report")
print("-" * 60)

print(classification_report(y_test, y_pred))

# -----------------------------------------------------------
# Confusion Matrix
# -----------------------------------------------------------

plt.figure(figsize=(6,6))

ConfusionMatrixDisplay.from_predictions(
    y_test,
    y_pred,
    cmap="Blues",
    values_format="d"
)

plt.title("Confusion Matrix")

plt.show()

# -----------------------------------------------------------
# ROC Curve
# -----------------------------------------------------------

fpr, tpr, _ = roc_curve(y_test, y_prob)

roc_auc = auc(fpr, tpr)

plt.figure(figsize=(7,6))

plt.plot(fpr, tpr, label=f"ROC AUC = {roc_auc:.4f}")
plt.plot([0,1],[0,1],'k--')

plt.xlabel("False Positive Rate")
plt.ylabel("True Positive Rate")
plt.title("ROC Curve")
plt.legend()

plt.show()

# -----------------------------------------------------------
# Precision Recall Curve
# -----------------------------------------------------------

precision, recall, _ = precision_recall_curve(y_test, y_prob)

ap = average_precision_score(y_test, y_prob)

plt.figure(figsize=(7,6))

plt.plot(recall, precision, label=f"AP = {ap:.4f}")

plt.xlabel("Recall")
plt.ylabel("Precision")
plt.title("Precision-Recall Curve")
plt.legend()

plt.show()

# -----------------------------------------------------------
# Feature Importance
# -----------------------------------------------------------

feature_names = preprocessor.get_feature_names_out()

importance = best_model.feature_importances_

importance_df = (
    pd.DataFrame({
        "Feature": feature_names,
        "Importance": importance
    })
    .sort_values(by="Importance", ascending=False)
)

print("\nTop 20 Important Features")
print("-" * 60)

print(importance_df.head(20))

plt.figure(figsize=(10,8))

plt.barh(
    importance_df.head(20)["Feature"][::-1],
    importance_df.head(20)["Importance"][::-1]
)

plt.title("Top 20 Feature Importances")

plt.tight_layout()

plt.show()

print("\nStage 11 Completed Successfully.")
```

```python
# ============================================================
# Stage 12 : Threshold Analysis & Best Model Selection
# ============================================================

import pandas as pd
import numpy as np

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    accuracy_score
)

print("=" * 70)
print("STAGE 12 : BEST MODEL SELECTION")
print("=" * 70)

# -----------------------------------------------------------
# Best Model
# -----------------------------------------------------------

final_model = best_models["CatBoost"]

# -----------------------------------------------------------
# Prediction Probabilities
# -----------------------------------------------------------

y_prob = final_model.predict_proba(X_test_processed)[:, 1]

# -----------------------------------------------------------
# Threshold Analysis
# -----------------------------------------------------------

thresholds = [0.10, 0.20, 0.30, 0.40, 0.50]

results = []

for threshold in thresholds:

    y_pred = (y_prob >= threshold).astype(int)

    results.append([
        threshold,
        accuracy_score(y_test, y_pred),
        precision_score(y_test, y_pred, zero_division=0),
        recall_score(y_test, y_pred, zero_division=0),
        f1_score(y_test, y_pred, zero_division=0)
    ])

threshold_df = pd.DataFrame(
    results,
    columns=[
        "Threshold",
        "Accuracy",
        "Precision",
        "Recall",
        "F1 Score"
    ]
)

print("\nThreshold Comparison")
print("-" * 60)
print(threshold_df)

# -----------------------------------------------------------
# Recommended Threshold
# -----------------------------------------------------------

best_row = threshold_df.loc[
    threshold_df["F1 Score"].idxmax()
]

print("\nRecommended Threshold")
print("-" * 60)

print(best_row)

print("\nFinal Selected Model : CatBoost")
```

```python
# ============================================================
# Stage 13 : Save Final Deployment Model
# ============================================================

import os
import joblib

from sklearn.pipeline import Pipeline

print("=" * 70)
print("STAGE 13 : SAVE FINAL MODEL")
print("=" * 70)

# -----------------------------------------------------------
# Create Final Pipeline
# -----------------------------------------------------------

final_pipeline = Pipeline([
    ("preprocessor", preprocessor),
    ("model", best_models["CatBoost"])
])

# -----------------------------------------------------------
# Fit Pipeline
# -----------------------------------------------------------

print("\nFitting Final Pipeline...")

final_pipeline.fit(X_train, y_train)

print("Pipeline Training Completed.")

# -----------------------------------------------------------
# Create Models Folder
# -----------------------------------------------------------

model_dir = "/content/drive/MyDrive/AI_Financial_Risk_Analytics/models"

os.makedirs(model_dir, exist_ok=True)

# -----------------------------------------------------------
# Save Model
# -----------------------------------------------------------

model_path = os.path.join(model_dir, "best_model.pkl")

joblib.dump(final_pipeline, model_path)

print("\nModel Saved Successfully!")

print(f"\nLocation:\n{model_path}")
```
